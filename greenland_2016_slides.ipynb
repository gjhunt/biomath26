## Models, hypotheses, and statistical tests

**Model → test hypothesis → discrepancy statistic → P value**

- A **model** mathematically describes data variability, assuming how data were collected, analyzed, and selected for reporting. Often there are some **parameters** of our model, say $\theta$, that we're interested in. 
- A **test hypothesis** typically specifies something about $\theta$, often $H:\theta=\theta_0$. Commonly use $H:\theta=0$ called a **null hypothesis**.
- Often use a **test statistic** to measure discrepancy between data and the hypothesis/assumptions. 
- Summarize test stat. with a **P value**: 

> “The P value is then the probability that the chosen test statistic would have been at least as large as its observed value if every model assumption were correct, including the test hypothesis.”
>
> — Greenland et al., p. 339


::: {.source}
Greenland et al. (2016), pp. 338–339; quotation p. 339.
:::

## A little notation: what varies, and what is assumed?

Let 

- **$D_\mathrm{obs}$** be the observed data,
- **$\theta$** the effect,
- **$A$** the other model assumptions,
- **$H$** the point hypothesis $H:\theta=\theta_0$,
- **T** the discrepancy statistic (larger values mean less agreement).
- **$D^*$**: other hypothetical data under H and A.

Then the $P$ value is:

$$P=\Pr\!\left\{T(D^*)\ge T(D_{\mathrm{obs}})\mid\theta=\theta_0,A\right\}.$$

This probability is over **hypothetical data**, with $\theta_0$ and $A$ held fixed.


## P values: the shaded tail areas {.full-figure-slide}

For example a $t$-test might be:

$$t=\frac{\bar{X}-\theta_0}{s/\sqrt{n}},\qquad
P=\Pr\left\{|t^*|\ge|t_{\mathrm{obs}}| \mid\theta=\theta_0,A\right\}$$


![](figures/p_value_tails.svg){fig-alt="Student t distribution with 19 degrees of freedom. Both tails beyond minus and plus 2.1 are shaded; their total probability is approximately 0.0493."}

The P value is the sum of both tail areas.


## The natural question: how certain is the hypothesis?

> “It is natural to express such certainty in terms of ‘probabilities’ of hypotheses.”
>
> — Greenland et al., p. 339

**We want to learn the effect size and assess how uncertain our estimate is.**

It is natural to want probabilities for possible effect sizes—but conventional frequentist calculations provide **probabilities of data patterns under assumed effect sizes and other model assumptions**.

$$\Pr(\text{data pattern}\mid H,A)\quad\neq\quad\Pr(H\mid\text{data}).$$

These are different questions. 

::: {.source}
Greenland et al. (2016), “Uncertainty, probability, and statistical significance,” p. 339; quotation p. 339.
:::

::: {.notes}
Read the quote as the motivation for statistical analysis, not a dismissal of probabilities of hypotheses. The distinction is what the conventional frequentist calculation provides. The display contrasts conditioning directions; it is not an equation defining a P value, which additionally involves a tail event. H is a specified hypothesis and A denotes other assumptions. The quoted source uses double quotation marks around probabilities; single marks here distinguish the inner quotation. Biological example and notation added.
:::

## The assumptions quietly treated as known

Traditional definitions of the P value emphasize the null hypothesis part, while

> “treating all other assumptions used to compute the P value as if they were known to be correct.”
>
> — Greenland et al., p. 339

Importantly:

> “this statistical model often incorporates unrealistic or at best unjustified assumptions.”
>
> — Greenland et al., p. 338

The full P value calculation is done assuming **H together with the other assumptions A**, including:

- Sampling, allocation, and dependence among observations.
- Measurement, missing data, and selection of analyses for reporting.

**The P value calculation assumes these conditions are satisfied; it does not establish that they are.** 

::: {.source}
Greenland et al. (2016), “Statistical models, hypotheses, and tests,” p. 338; “Uncertainty, probability, and statistical significance,” p. 339; quotation p. 339.
:::

## P values: compatibility with the entire model

> “Recognizing that these other assumptions are often questionable if not unwarranted, we will adopt a more general view of the P value as a statistical summary of the compatibility between the observed data and what we would predict or expect to see if we knew the entire statistical model (all the assumptions used to compute the P value) were correct.”
>
> — Greenland et al., p. 339

**The reference is the entire model: H together with A.**

A small P flags a discrepancy with that joint model. It cannot distinguish a false test hypothesis from a failure of another assumption.

::: {.source}
Greenland et al. (2016), “Uncertainty, probability, and statistical significance,” p. 339; quotation p. 339.
:::

::: {.notes}
This quotation follows immediately after the warning about treating other assumptions as known. Read the conditional ending carefully: compatibility is assessed against what the model would predict if all its assumptions were correct. The statistic does not validate those assumptions. A small P may arise from a false test hypothesis, violated protocols, or selection for a small result. A large P does not establish the model either. Compatibility here is not a posterior probability of a hypothesis.
:::

## Statistical significance and surprisal

“Statistically significant” conventionally means **P ≤ α**, usually α = 0.05.

> “the arbitrary classification of results into ‘significant’ and ‘non-significant’ is unnecessary for and often damaging to valid interpretation of data”
>
> — Greenland et al., p. 338 (in most scientific settings)

An added way to express P continuously is **surprisal**, in bits:

$$S=-\log_2(P).$$

| P | 0.50 | 0.05 | 0.01 | 1E-10 | 
|---|---:|---:|---:|---:|
| S (bits) | 1 | 4.32 | 6.64 | 33.2 | 

**Coin calibration:** s bits corresponds to the probability of **all heads in s specified independent fair-coin flips**: $P=2^{-s}$.
For example, **P = 1/32 gives S = 5 bits**, like five heads in five flips.

Larger S means greater surprise under the **joint model**; it is not a probability that the hypothesis is true.

::: {.source}
Greenland et al. (2016), pp. 338–339. Added connection: [Rafi & Greenland (2020), “Semantic and cognitive tools…”](https://pmc.ncbi.nlm.nih.gov/articles/PMC7528258/), BMC Medical Research Methodology 20:244.
:::

::: {.notes}
The significance critique is from the 2016 paper; surprisal is an explicitly added connection to later work by Rafi and Greenland (2020). Alpha is fixed in advance, whereas P is computed from the observed data. S is a monotonic re-expression of P and inherits its model and selection assumptions. Under a properly calibrated P value, the bits provide a scale for the surprise of the tail event, not a posterior probability or a Bayes factor. The coin analogy concerns a specified sequence in a fixed number of flips, not the waiting time until a run of heads. For noninteger S, such as 4.32, interpret it as a position on a continuous information scale, between the integer-flip examples. The analogy calibrates the probability of the tail event, not the exact observed data. S is not a count of observed flips and adds no information beyond P. The P=.05 threshold has no special discontinuity on this scale. Do not interpret the transform as evidence exclusively against H rather than against the joint assumptions.
:::

## From tests to estimates to Conf Intervals

Test a range of candidate effects, rather than only θ = 0, and collect all values $\theta$ where $P$ is large. 

For example, collect $\theta$ where $P > \alpha$ (e.g. $\alpha=0.05$) then we get a (e.g.) **95% confidence interval**:

$$C =\{\theta: \text{P value under $\theta$ } > 0.05\}.$$


**Read the range:** Effects in the interval are reasonably compatible with the data. Effects outside the interval are less compatible with the data. 

![ ](figures/ci_compatibility.svg){fig-alt="P-value curve for a normal estimate of 3 with known SE 2. Candidate effects from approximately minus 0.92 to 6.92 have P values above 0.05."}


## 1. P is the probability the hypothesis is true

**Misconception:** P = 0.01 means the null has a 1% probability of being true.

**Correction:** P describes discrepancies assuming the hypothesis and other model assumptions. It does not assign a probability to the hypothesis.

**Remember:** Keep the direction of conditioning straight.

::: {.source}
Greenland et al. (2016), misconception 1, p. 340. Brief paraphrase.
:::

::: {.notes}
Relate this to the earlier probability-of-data versus probability-of-hypothesis slide.
:::

## 2. P is the probability that chance caused the result

**Misconception:** P = 0.08 means an 8% chance that chance alone produced the association.

**Correction:** The calculation already assumes a model of chance variation, including the test hypothesis. It does not give the probability that this explanation is correct.

**Remember:** A probability derived from assumptions is not a probability of those assumptions.

::: {.source}
Greenland et al. (2016), misconception 2, p. 340. Brief paraphrase.
:::

::: {.notes}
Chance variation can coexist with real effects; avoid an either-chance-or-effect story.
:::

## 3. Significance establishes that the hypothesis is false

**Misconception:** P ≤ 0.05 establishes that the tested hypothesis is false and must be rejected.

**Correction:** A small P flags unusual data under the joint model. Random variation, a false hypothesis, or another failed assumption can produce it.

**Remember:** Investigate the whole model before attributing the discrepancy to H.

::: {.source}
Greenland et al. (2016), misconception 3, p. 341. Brief paraphrase.
:::

::: {.notes}
A prespecified decision rule can reject at alpha, but that procedural decision is not proof of falsity or an automatic scientific conclusion.
:::

## 4. Nonsignificance establishes that the hypothesis is true

**Misconception:** P > 0.05 means the test hypothesis should be accepted as true.

**Correction:** The observed statistic is not unusual under that model. Many other effect sizes may also fit these data.

**Remember:** Failure to reject is not confirmation.

::: {.source}
Greenland et al. (2016), misconception 4, p. 341. Brief paraphrase.
:::

::: {.notes}
A large P can also result from imprecision or selection for nonsignificance.
:::

## 5. A large P favors the tested hypothesis

**Misconception:** A large P by itself is evidence favoring the tested hypothesis over alternatives.

**Correction:** A large P may simply reflect an inability to distinguish effects. Other hypotheses can have even larger P values under the same testing framework.

**Remember:** Assess the alternatives and the range of compatible effects directly.

::: {.source}
Greenland et al. (2016), misconception 5, p. 341. Brief paraphrase.
:::

::: {.notes}
This is the paper’s compatibility argument; it is not a claim that evidence for null effects is impossible under all inferential frameworks.
:::

## 6. P > 0.05 means no effect was observed

**Misconception:** A nonsignificant null test demonstrates no association or effect.

**Correction:** The estimated effect can be far from zero even when its P value exceeds 0.05. The estimate and interval show what was observed and what remains unresolved.

**Remember:** Report the estimated effect instead of replacing it with “no effect.”

::: {.source}
Greenland et al. (2016), misconception 6, p. 341. Brief paraphrase.
:::

::: {.notes}
Distinguish an observed association from a causal effect and from a population parameter.
:::

## 7. Statistical significance means scientific importance

**Misconception:** A significant result identifies a biologically important relationship.

**Correction:** With enough precision, tiny effects—or small assumption violations—can produce small P values.

**Remember:** Judge magnitude in biological units and against a meaningful effect size.

::: {.source}
Greenland et al. (2016), misconception 7, p. 341. Brief paraphrase.
:::

::: {.notes}
An effect estimate and its interval are needed to assess importance.
:::

## 8. Nonsignificance means a small effect

**Misconception:** If a result is not significant, its effect must be small.

**Correction:** Large effects may remain unresolved in noisy or small studies. A large P does not bound the effect size.

**Remember:** Check whether the interval still includes important effects.

::: {.source}
Greenland et al. (2016), misconception 8, p. 341. Brief paraphrase.
:::

::: {.notes}
This differs from item 6: the error here is inferring small magnitude, rather than claiming that no association was observed.
:::

## 9. P is the probability of exactly these data

**Misconception:** P is the probability of obtaining precisely the observed result under the hypothesis.

**Correction:** P includes the observed discrepancy and discrepancies at least as extreme, according to the chosen statistic.

**Remember:** Think tail area, not density at one point.

::: {.source}
Greenland et al. (2016), misconception 9, p. 341–342. Brief paraphrase.
:::

::: {.notes}
Use the shaded t-distribution figure. For continuous data, the probability of an exact point is zero; the density at a point is not its P value.
:::

## 10. A 5% test makes a discovery 95% reliable

**Misconception:** Rejecting at α = 0.05 means a 5% probability this finding is a false positive.

**Correction:** Alpha controls rejection frequency across repeated valid tests when the null is true. It does not give the probability that a particular discovery is false.

**Remember:** Error rates condition on hypotheses, not on this observed discovery.

::: {.source}
Greenland et al. (2016), misconception 10, p. 342. Brief paraphrase.
:::

::: {.notes}
The mix of true and false null hypotheses and the power under alternatives affect false-discovery proportions.
:::

## 11. P = 0.05 and P ≤ 0.05 convey the same information

**Misconception:** Reporting an exact borderline value is equivalent to reporting only significance.

**Correction:** P ≤ 0.05 groups P = 0.05 with values such as P = 0.0001, which convey very different degrees of incompatibility.

**Remember:** Keep the numerical result rather than just the threshold label.

::: {.source}
Greenland et al. (2016), misconception 11, p. 342. Brief paraphrase.
:::

::: {.notes}
This concerns information lost in an inequality; it is distinct from item 12’s broader reporting recommendation.
:::

## 12. Inequalities are the best way to report P

**Misconception:** Reporting only P < 0.05 or P > 0.05 is adequate.

**Correction:** These labels hide potentially important differences, such as P = 0.06 versus P = 0.70. Report numerical P values to sensible precision.

**Remember:** Very small values may reasonably be reported with an appropriate bound.

::: {.source}
Greenland et al. (2016), misconception 12, p. 342. Brief paraphrase.
:::

::: {.notes}
The paper allows inequalities for very small P values where further numerical precision is not useful or reliable.
:::

## 13. Significance belongs to the biological phenomenon

**Misconception:** A statistical test detects whether a biological effect is “significant.”

**Correction:** Statistical significance labels a test result at a chosen cutoff. It depends on study size, uncertainty, assumptions, and the analysis.

**Remember:** Describe the biological effect and the statistical result separately.

::: {.source}
Greenland et al. (2016), misconception 13, p. 342. Brief paraphrase.
:::

::: {.notes}
Avoid statements that treat significance as a stable intrinsic property of a population effect.
:::

## 14. Tests must always be two-sided

**Misconception:** A one-sided P value is always inappropriate.

**Correction:** A one-sided test can match a genuinely one-sided scientific question, such as whether an effect exceeds a specified boundary.

**Remember:** State and justify the direction; do not choose it after seeing the result.

::: {.source}
Greenland et al. (2016), misconception 14, p. 342. Brief paraphrase.
:::

::: {.notes}
The paper uses an at-least-as-good-as-standard treatment question. The relevant boundary or noninferiority margin must be scientifically defined.
:::

## 15. Mostly nonsignificant studies support no effect

**Misconception:** If none or few of the studies are significant, the literature supports the null.

**Correction:** Combining information can reveal an association that individual imprecise studies do not resolve.

**Remember:** Synthesize estimates with attention to bias and heterogeneity; do not count significance labels.

::: {.source}
Greenland et al. (2016), misconception 15, p. 342–343. Brief paraphrase.
:::

::: {.notes}
The paper illustrates this with five P values of .10 combined by Fisher’s method, giving about .01 under its independence and validity assumptions.
:::

## 16. Opposite significance labels mean conflicting effects

**Misconception:** P < 0.05 in one study and P > 0.05 in another establish disagreement.

**Correction:** The paper’s trials both estimate 3, but SEs 2 and 1 give P ≈ 0.13 and 0.003. The estimated effects agree exactly.

**Remember:** Estimate and test the difference between effects directly.

::: {.source}
Greenland et al. (2016), misconception 16, p. 343. Brief paraphrase.
:::

::: {.notes}
For these independent normal estimates the observed contrast is zero and its P value is one; this does not prove equality of population effects.
:::

## 17. Identical P values mean agreeing effects

**Misconception:** The same P value in two studies establishes agreement.

**Correction:** Estimates 3 (SE 1) and 12 (SE 4) both give P ≈ 0.003. Their direct contrast is 9, with P ≈ 0.03.

**Remember:** P values alone do not reveal agreement in effect size.

::: {.source}
Greenland et al. (2016), misconception 17, p. 343. Brief paraphrase.
:::

::: {.notes}
Paper example with independent normal estimates and known SEs. Contrast SE is sqrt(17); two-sided P is about .029.
:::

## 18. A small P predicts an equally small replication P

**Misconception:** A small P makes an equally small P likely in the next study.

**Correction:** An observed P is not a replication probability. Future P values depend on the true effect, study design, assumptions, and sampling variation.

**Remember:** Compare estimates and uncertainty, rather than demanding replication of a P value.

::: {.source}
Greenland et al. (2016), misconception 18, p. 343. Brief paraphrase.
:::

::: {.notes}
Under independent continuous valid tests of a true null, the next P is uniform; observing .03 does not make another P at most .03 more likely than 3%.
:::

## 19. This interval has a 95% chance of containing θ

**Misconception:** The observed frequentist 95% interval assigns 95% probability to containing the true effect.

**Correction:** The confidence level describes coverage of intervals across repeated valid studies. It does not by itself give a posterior probability for this interval.

**Remember:** A credible interval answers a different question using additional assumptions, including a prior.

::: {.source}
Greenland et al. (2016), misconception 19, p. 343–344. Brief paraphrase.
:::

::: {.notes}
Parameter fixed, interval varying in the coverage calculation. A posterior interpretation may coincide numerically under particular additional assumptions but does not follow from coverage alone.
:::

## 20. Values outside the CI have been disproved

**Misconception:** Every effect outside the 95% interval has been ruled out by the data.

**Correction:** Interval limits depend on assumptions and a chosen confidence level. Just outside and just inside an endpoint can be almost equally compatible.

**Remember:** Treat the endpoints as a reporting convention, not a boundary of possibility.

::: {.source}
Greenland et al. (2016), misconception 20, p. 344. Brief paraphrase.
:::

::: {.notes}
Return to the compatibility curve. Model misspecification may alter both the curve and the interval.
:::

## 21. Overlapping CIs mean no significant difference

**Misconception:** If two 95% intervals overlap, their effects cannot differ significantly.

**Correction:** The paper’s intervals [1.04, 4.96] and [4.16, 19.84] overlap, yet the direct comparison gives P ≈ 0.03.

**Remember:** Calculate the contrast and its uncertainty, including covariance where needed.

::: {.source}
Greenland et al. (2016), misconception 21, p. 344. Brief paraphrase.
:::

::: {.notes}
The example uses independent normal estimates 3 and 12, with known SEs 1 and 4. The P value is about .029.
:::

## 22. A 95% CI contains 95% of future estimates

**Misconception:** This interval predicts where 95% of estimates from future studies will fall.

**Correction:** Coverage concerns the true effect, not future point estimates. In an ideal equal-SE normal example, repeated-pair inclusion is about 83%, not 95%.

**Remember:** Use a prediction calculation when the target is a future estimate.

::: {.source}
Greenland et al. (2016), misconception 22, p. 344. Brief paraphrase.
:::

::: {.notes}
For independent unbiased normal estimates of the same fixed effect with equal known SE s, their difference has SD sqrt(2)s. Unconditional inclusion is 2 Phi(1.96/sqrt(2)) − 1 ≈ .834; it is not a universal conditional probability for a specific observed interval.
:::

## 23. Excluding zero makes an interval more precise

**Misconception:** The interval excluding zero is necessarily more precise.

**Correction:** Precision is reflected by interval width on the appropriate scale. Crossing the null is a separate issue.

**Remember:** Compare widths: [5, 40] is wider than [−5, 10], despite excluding zero.

::: {.source}
Greenland et al. (2016), misconception 23, p. 344. Brief paraphrase.
:::

::: {.notes}
The paper has an arithmetic slip: 5 to 40 is 35 units wide, not 30. The second interval is 15 units wide. This slide avoids repeating the paper’s incorrect width and half-width claim.
:::

## 24. 90% power makes a null conclusion 90% reliable

**Misconception:** With 90% power, accepting the null after P > 0.05 has only a 10% chance of error.

**Correction:** The 10% is the frequency of nonrejection when a specified alternative is true, under the design assumptions. It is not a probability that this null conclusion is wrong.

**Remember:** Power describes a design at an alternative, not the truth of an observed conclusion.

::: {.source}
Greenland et al. (2016), misconception 24, p. 345. Brief paraphrase.
:::

::: {.notes}
Define power before discussing it: probability before the study that the chosen testing rule rejects under a specified effect and model. It depends on the alternative, sample size, variability, and decision rule.
:::

## 25. Nonsignificance plus high power favors the null

**Misconception:** P > 0.05 and 90% power at an alternative establish support for the null over that alternative.

**Correction:** The observed data can be more compatible with the alternative despite a nonsignificant null test. Planned power does not compare the two hypotheses against these data.

**Remember:** Evaluate the null and meaningful alternatives directly using comparable measures.

::: {.source}
Greenland et al. (2016), misconception 25, p. 345. Brief paraphrase.
:::

::: {.notes}
Added optional counterexample: known SE 1, two-sided alpha .05, design alternative 3.24 gives about 90% power. An estimate 1.8 yields P about .072 under zero and .150 under 3.24. Plug-in observed power is not independent evidence.
:::

## Assumptions include the analysis choices

> “All statistical methods (whether frequentist or Bayesian, or for testing or estimation, or for inference or decision) make extensive assumptions about the sequence of events that led to the results presented—not only in the data generation, but in the analysis choices.”

**For critical evaluation, describe the full path to the result:**

- Study motivation and design.
- Original analysis plan and inclusion/exclusion criteria.
- All analyses conducted.

::: {.source}
Greenland et al. (2016), closing guideline (g), p. 347. Quotation excerpt; reporting points paraphrased. [Paper](https://pmc.ncbi.nlm.nih.gov/articles/PMC4877414/).
:::


## ASA (2016): using P values responsibly

1. P-values can indicate how incompatible the data are with a specified statistical model.
2. P-values do not measure the probability that the studied hypothesis is true, or the probability that the data were produced by random chance alone.
3. Scientific conclusions and business or policy decisions should not be based only on whether a p-value passes a specific threshold.
4. Proper inference requires full reporting and transparency.
5. A p-value, or statistical significance, does not measure the size of an effect or the importance of a result.
6. By itself, a p-value does not provide a good measure of evidence regarding a model or hypothesis.

Also (my own opinion): present P values as continuous measures of incompatibility. 

::: {.source}
Wasserstein & Lazar (2016), ASA Statement, principles 1–6. Paraphrased summary. [Statement](https://doi.org/10.1080/00031305.2016.1154108); [ASA summary](https://www.amstat.org/asa/files/pdfs/p-valuestatement.pdf).
:::
